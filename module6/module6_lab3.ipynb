```python
import pandas as pd
import sqlite3
import seaborn as sns
import matplotlib.pyplot as plt
import scipy
```

```python
conn = sqlite3.connect("custtable.db")
c = conn.cursor()
```

```python
pd.read_sql("SELECT * FROM customer_information", conn)
```

```python
pd.read_sql("SELECT * FROM customer_information WHERE Income >= 75000", conn)
```

```python
pd.read_sql("SELECT * FROM customer_information ORDER BY Income DESC", conn)
```

```python
pd.read_sql("SELECT * FROM customer_information WHERE Name LIKE 'S%'", conn)
```

```python
table1df = pd.read_sql("SELECT * FROM customer_information ORDER BY Income DESC", conn)
```

```python
sns.barplot(data=table1df, x="Name", y="Income")
plt.title("Alderman Leads Income Just Ahead of Coelho")
plt.xlabel("Customer Name")
plt.ylabel("Annual Income ($)")
plt.xticks(rotation=45)
plt.show
```

```python
pd.read_sql("SELECT City, AVG(Income) AS Avg_Income FROM customer_information GROUP BY City ORDER BY Avg_Income DESC", conn)
```

```python
table2df = pd.read_sql("SELECT City, AVG(Income) AS Avg_Income FROM customer_information GROUP BY City ORDER BY Avg_Income DESC", conn)
```

```python
sns.barplot(data=table2df, x="City", y="Avg_Income")
plt.title("Anaheim Leads Average Income By Small Margin")
plt.xlabel("City Name")
plt.ylabel("Annual Income ($)")
plt.xticks(rotation=45)
plt.show
```
