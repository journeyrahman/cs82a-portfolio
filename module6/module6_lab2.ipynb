```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("custtable.db")
c = conn.cursor()
```

```python
c.execute("DROP TABLE IF EXISTS customer_information")
```

```python
c.execute("""CREATE TABLE customer_information (
            Customer_ID integer PRIMARY KEY,
            Name text,
            City text,
            Income real)""")
```
