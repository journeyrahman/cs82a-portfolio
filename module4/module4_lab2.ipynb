```python
round(booksdf.describe(),3)
```

Based on the values for each quartile, the "weight" column is likely to be distributed evenly, overall.

```python
print(booksdf["weight"].mean())
print(booksdf["weight"].median())
print(booksdf["weight"].std())
```

The mean "weight" of the dataset is slightly below the median. This likely indicates that there are a few rows with an especially low "weight" value. The standard deviation is approximately a third of the median, so the average distance of each data point from the mean is not huge.

```python
booksdf.groupby(["cover"]).mean()
```

When we separate the "weight" values by "cover" type, we see that the mean is quite different between the two groups. The "hb" group's mean "weight" is higher by almost 180--close to two-thirds of the standard deviation of "weight" for the whole dataset. This explains the earlier mean being a bit lower than the median: The "pb" rows have much lower "weight", and there may be more "pb" rows than "hb" rows.

```python
booksdf.sort_values(by=["cover"])
```

We see here that there are 7 "hb" rows and 8 "pb" rows. There is indeed one more "pb" row" than "hb", which brings down the mean "weight" for the dataset.</p>
Additionally, the median "cover" value is "pb", as that is the middle value in the above list that is ordered by "cover".
