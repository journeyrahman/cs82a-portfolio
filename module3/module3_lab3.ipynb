```python
import pandas as pd
bridgesdf = pd.read_csv("bridges.csv")
```

The column of interest in this dataset is length. Assuming that this column is measured in feet, a plausible range would be between ~0.2 miles (1000 feet) and ~0.75 mile (4000 feet). The reasoning for the lower bound is that, if a waterway is sufficiently narrow, the land surrounding it is often redeveloped to avert the flow entirely out of a highly-trafficked area. As for the upper bound, bridges usually are not a full mile long because such a large building project can be more costly than exploring an alternative, such as providing ferry service across a wide body of water. The dataset will be screened in accordance with this assumption.<p>

"length" lower bound: 1000 <br>
"length" upper bound: 4000

We can see that there are datapoints that lie outside our assumed bounds. Let us identify the outliers for "length" so that they can be trimmed out of the dataset.

```python
bridgesdf [ (bridgesdf ["length"] < 1000) | (bridgesdf ["length"] > 4000) ]
```

We can see that exactly one "length" outlier is above 4000. The other 12 returned rows are below 1000, but some are only barely below 1000, such as "Id"==E7, which has "length"==990. The rows with "length"<1000 are distributed evenly and none of them are excessively below "length"==1000, so they represent genuine and relevant data points. For this reason, we will not consider them outliers, as excluding them from the data would artificially skew any subsequent analysis.

<p>We will only screen the "length"==4558 outlier out of the data by dropping that row from the dataset, as it is an isolated value far above the rest.

```python
bridgesdf.drop([31])
```

```python
bridgesdf = bridgesdf.drop([31])
```

```python
bridgesdf.shape
```

```python
bridgesdf [ (bridgesdf ["length"] > 4000) ]
```

The outlier with "length">4000 has been dropped from the dataset. Let's create a new csv so that the cleaned data is ready for use.

```python
bridgesdf.to_csv("bridges_cleaned.csv")
```
