````python

# Import pandas for data manipulation, seaborn for visualization,
# and matplotlib for customizing the chart.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Load the CSV file
# --------------------------------------------------

# Replace "penguins.csv" with the path to your CSV file.
df = pd.read_csv("penguins.csv")

# Display the first five rows to verify that the data loaded correctly.
print("First five rows:")
print(df.head())


# --------------------------------------------------
# 2. Summarize missing data
# --------------------------------------------------

# Count the missing (NaN) values in each column.
missing_count = df.isnull().sum()

# Calculate the percentage of missing values in each column.
missing_percent = (df.isnull().mean() * 100).round(2)

# Combine the counts and percentages into a readable summary table.
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percent": missing_percent
})

# Display the summary, including columns with no missing values.
print("\nMissing Data Summary:")
print(missing_summary)


# --------------------------------------------------
# 3. Compare flipper length across species
# --------------------------------------------------

# Group the data by species and calculate descriptive statistics.
# dropna() excludes rows without a flipper length from this comparison.
flipper_summary = (
    df.dropna(subset=["flipper_length_mm"])
      .groupby("species")["flipper_length_mm"]
      .agg(["count", "mean", "median", "std", "min", "max"])
      .round(2)
)

# Rename columns to make the results easier to interpret.
flipper_summary.columns = [
    "Count", "Mean (mm)", "Median (mm)",
    "Std Dev (mm)", "Min (mm)", "Max (mm)"
]

# Print the statistics so you can compare typical lengths and variability.
print("\nFlipper Length Summary by Species:")
print(flipper_summary)


# --------------------------------------------------
# 4. Produce one labeled chart
# --------------------------------------------------

# Use a clean seaborn theme for a readable chart.
sns.set_theme(style="whitegrid")

# Create a boxplot to compare the distribution of flipper lengths
# for each species. Each box shows the median and interquartile range.
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="species",
    y="flipper_length_mm",
    hue="species",
    palette="Set2",
    legend=False
)

# Add a descriptive title and axis labels, including measurement units.
plt.title("Flipper Length Distribution by Penguin Species",
          fontsize=14, pad=12)
plt.xlabel("Penguin Species", fontsize=11)
plt.ylabel("Flipper Length (mm)", fontsize=11)

# Keep the species labels readable and prevent unnecessary clutter.
plt.tight_layout()

# Display the finished chart.
plt.show()
````

````python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load the penguin dataset.
df = pd.read_csv("penguins.csv")

# 2. Remove rows with missing flipper lengths, island names,
#    or species names so the comparison uses complete observations.
penguins_clean = df.dropna(
    subset=["flipper_length_mm", "island", "species"]
)

# 3. Group by island and species, then calculate descriptive statistics.
#    These values are calculated from your CSV when you run the code.
summary = (
    penguins_clean
    .groupby(["island", "species"])["flipper_length_mm"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .round(2)
)

# 4. Print the summary table.
print("Flipper Length by Island and Species:")
print(summary)

# 5. Create a grouped boxplot to compare distributions.
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=penguins_clean,
    x="island",
    y="flipper_length_mm",
    hue="species",
    palette="Set2"
)

# 6. Add a descriptive title and axis labels.
plt.title("Penguin Flipper Length by Island and Species")
plt.xlabel("Island")
plt.ylabel("Flipper Length (mm)")
plt.legend(title="Species")

# 7. Adjust spacing and display the chart.
plt.tight_layout()
plt.show()
````
