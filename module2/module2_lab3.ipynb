<p>For the purposes of this lab, the following cut-offs will be used for categorizing temperature:</p>

<p>"Hot" weather is above 25C. <br>
"Warm" weather is above 20C through 25C. <br>
"Cool" weather is above 15C through 20C. <br>
"Cold" weather is 15C and below. <br></p>

```python
temps = [10, 16, 23, 28, 15, 22, 30]
```

```python
def tf(temps):
    for n in temps:
        if n > 25:
            print(n,"degrees Celsius is hot.")
        elif n > 20:
            print(n,"degrees Celsius is warm.")
        elif n > 15:
            print(n,"degrees Celsius is cool.")
        else:
            print(n,"degrees Celsius is cold.")
```

```python
tf(temps)
```
