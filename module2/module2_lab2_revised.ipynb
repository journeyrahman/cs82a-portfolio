```python
a, b, c = 0, 0, 8
while (b <= c):
    b = b + 1
    print (a)
    a += 5
```

```python
def multwhile(c):
    a, b = 0, 0
    while (b <= c):
        b = b + 1
        a += 5
        if b == c:
            print(a)

multwhile(8)
```

```python
print(5 * 8)
```

```python
multwhile(6)
```

```python
print(5 * 6)
```

The while loop terminates after the instruction to "add 5 to a" (a += 5) iterates 8 times. This is because b exists solely as a counter for how many times the loop iterates, and the function allows us to modify the number of iterations as desired by plugging in different values for c.

I would like to provide credit to GeeksForGeeks.org for modeling how to embed a counter that iterates:
https://www.geeksforgeeks.org/python/loops-in-python/

```python
times_tables = [0]
n = 8
for i in range (0, n):
    times_tables.append(5)
print(times_tables)
print(sum(times_tables))
```

```python
def multfor(n):
    times_tables = [0]
    for i in range (0, n):
        times_tables.append(5)
    print(times_tables)
    print(sum(times_tables))

multfor(8)
```

```python
print(5 * 8)
```

```python
multfor(6)
```
